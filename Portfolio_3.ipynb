<h1>Caso de uso: Canciones de Spotify<h1>

A continuaciónse obtendrá la información de un archivo JSON que contine un total de 45.128 canciones de spotify, cada una de ellas con la siguiente información
- *track*: nombre de la canción
- *album*: nombre del album al que pertenece
- *artist*: nombre del artista
- *duration*: duración en milisegundos
- *genres*: lista de generos del artista

```python
import pandas as pd
```

```python
df_canciones = pd.read_json(r'..\Troncal\T1-herramientas_programación\ejercicios\data\tracks.json')
```

<h3>Pregunta<h3>
<h4>
1. ¿Qué duración tiene la canción `The Show Must Go On - Remastered 2011` del grupo `Queen`? <h4>

```python
df_canciones[(df_canciones.artist=="Queen") & (df_canciones.track=="The Show Must Go On - Remastered 2011")]
```

<h4>2. ¿Cuántas canciones diferentes a remasterizado Queen? Es decir, ¿Cuántas de sus canciones contienen el texto Remastered?

```python
df_canciones[(df_canciones.artist=="Queen") & (df_canciones.track.str.contains("Remastered"))].track.count()
```

```python
df_canciones[(df_canciones.artist=="Queen") & (df_canciones.track.str.contains("Remastered"))].track.nunique()
```

<h4>3. ¿Cuantos artistas diferentes hay en los datos? ¿Y canciones diferentes?<h4>

```python
print("Hay", df_canciones.track.nunique(), "canciones y", df_canciones.artist.nunique(), "artistas.")
```

<h4>4. ¿Cuál es la canción con más géneros? ¿Y la que menos?

```python
#Utilizando map
(
    df_canciones
    .assign(
        n_generos=df_canciones.genres.map(lambda x: len(x))
    )
    .nlargest(1, "n_generos")
)
```

```python
#Utilizando apply
(
    df_canciones
    .assign(
        n_generos=df_canciones.genres.apply(lambda x: len(x))
    )
    .nsmallest(1, "n_generos")
)
  
```

<h4>5. ¿Cuántos segundos de duración suman todas las canciones? <h4>

```python
#Duración estaba en milisegundos, lo pasamos a segundos
df_canciones.duration.sum()/1000
```

<h4>6. ¿Qué artista ha compuesto más canciones?

```python
#Canciones únicas + duración total por artista
(
    df_canciones
    .groupby("artist")
    .agg(
        n_canciones=("track", "nunique"),
        duracion_total=("duration", "sum")
    )
    .sort_values("duracion_total", ascending=False)      
)
```

```python
#Otra opción, nº registros por artista
(
    df_canciones
    .groupby("artist")
    .agg(
        n_canciones=("track", "count"),
        
    )
    .sort_values("n_canciones", ascending=False)      
).head(3)
```

```python
#Cuantos duplicados hay de canciones por artista
df_canciones.duplicated(subset=["artist", "track"]).sum()
```

<h4>7. ¿Qué album contiene más canciones? ¿Y más artistas diferentes?<h4>

```python
 (
    df_canciones
    .groupby("album")
    .agg(
        n_canciones=("track", "nunique"),
    )
    .sort_values("n_canciones", ascending=False)
    .head(3)
    )
```

```python
(
    df_canciones
    .groupby("album")
    .agg(
        n_artistas=("artist", "nunique"),
    )
    .sort_values("n_artistas", ascending=False)
    .head(3)
    )
```

<h4>8.¿Qué album tiene mayor duración?<h4>

```python
(
    df_canciones
    .groupby("album")
    .agg(
        n_album=("duration", "sum"),
    )
    .sort_values("n_album", ascending=False)
    .head(3)
    )
```

<h4>9. ¿Qué artista compone, en media, las canciones más largas?¿Y las más cortas?

```python
(
    df_canciones
    .groupby("artist")
    .agg(
        cancion_larga=("duration", "mean"),
    )
    .nlargest(3, "cancion_larga")
    )
```

```python
(
    df_canciones
    .groupby("artist")
    .agg(
        cancion_larga=("duration", "mean"),
    )
    .nsmallest(3, "cancion_larga")
    )
```

<h4>10. ¿Qué porcentaje de las canciones son de género rock?<h4 >

```python
(
    df_canciones
    .assign(
        genero_rock=df_canciones.genres.apply(lambda x: "rock" in x)
    )
    .agg(
        {"genero_rock": lambda s: s.sum()/s.count()}
    )
    
)
```

```python
# option 2
(
    df_canciones
    .assign(
        is_rock=lambda df: df.genres.map(lambda genres: "rock" in genres)
    )
    .agg(
        {"is_rock": lambda s: s.sum() / s.count()}
    )
)
```
