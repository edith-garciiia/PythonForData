1. Sabiendo que los datos son de 2024, crear una columna `birth_year` que sea el año de nacimiento, calculado como 2024 menos la edad de cada jugador.

2. Sobreescribir la columna `release_clause` para que indique el valor como tipo numérico en millones.
    - *Nota*: algunos valores están en millones (€10M) y otros en miles (€10K).
  
3. Aplica el mismo método que has realizado para `release_clause` para `value` y `wage`.

4. Crear una columna que sea `effort`, calculada como la diferencia de la columna `potential` menos `overall`. ¿Qué jugadores están en el top 10 respecto a `effort`?

5. Transforma la columna `height` y `weight` en números flotantes, indicando respectivamente la altura en metros y el peso en kilogramos.

    - **Nota 1**: 1 pie son `30.48` centímetros
    - **Nota 2**: 1 libra son `0.453592` kilogramos

```python
def price_to_number(x):
    if x[-1] == 'K':
        return float(x[1:-1])/1000
    if x[-1] == 'M':
        return float(x[1:-1])
    else:
        return float(x[1:])/1000000
def feet_to_cm(x):
    return float(x.replace("'", ".")) * 30.48
def pounds_to_kg(x):
    return float(x.replace("lbs", "")) * 0.453592
    
```

```python
df_fifa=( 
    df_fifa
        .assign(
            birth_year=lambda df: 2024 - df['age'],
            release_clause= lambda df: df.release_clause.map(price_to_number),
            wage= lambda df: df.wage.map(price_to_number),
            value= lambda df: df.value.map(price_to_number),
            effort= lambda df: df.potential - df.overall,
            height= lambda df: df.height.map(feet_to_cm),
            weight= lambda df: df.weight.map(pounds_to_kg)
        )
)
```

**Ejercicio**: Crea una columna `overall_category` que asigne:
- Mediocre si el overall es menor a 60.
- Bueno si está entre 60 y 80.
- Elite si es mayor a 80.

```python
df_fifa=(
    df_fifa
    .assign(
        overall_category= lambda df: df.overall.map(
            lambda x: 'Mediocre' if x < 60 else
                      'Bueno' if x < 80 else 
                      'Elite' 
        )

    )
)
df_fifa.head()
```

**Ejercicio**: Crea una nueva columna llamada `age_group` que clasifique a los jugadores en:

Uno de los problemas más habituales es que nuestros datos están divididos en distintas tablas aun representando el mismo modelo y existe información implícita en la representación, por ejemplo en el nombre de las tablas o los ficheros.

Éste es el caso del dataset `baby-names`, que contiene un estudio sobre la frecuencia de los nombres que se les ponen a los bebes en USA en distintos años. Este dataset viene en un formato que proporciona un fichero por año, donde el año viene especificado en el nombre del fichero.

Este dataset tiene las variables `year`, `rank`, `name`, `frequency` y `sex`, pero la variable year está implícita en los datos (nombre del fichero) y no aparece en el modelo.

Crear una función que se le pase una lista de _paths_, que lea cada uno de los `DataFrames` a partir de cada uno de los paths, y que utilize `pd.concat` para devolver un único `DataFrame`. Además, cada dataframe debe incluir una nueva columna llamada `year` que indique el año representado en el propio nombre (path) del fichero. Para ello será necesario usar la función de strings `split`.

```python
display( pd.read_csv(r"C:\Users\edith\OneDrive\Documentos\Universidad\Máster IA\T1-herramientas_programación\notebooks\data\2014-baby-names-raw.csv").head(2) )
display( pd.read_csv(r"C:\Users\edith\OneDrive\Documentos\Universidad\Máster IA\T1-herramientas_programación\notebooks\data\2015-baby-names-raw.csv").head(2) )
```

```python
def read_by_year(path):
    name = path.split("/")[-1]
    year = name[:4]
    return (
        pd.read_csv(path)
        .assign(year=year)
    )
def read_baby_names(paths):
    df = (
        pd.concat(
            [read_by_year(path) for path in paths]
        )

    )
    return df

paths = [r"C:\Users\edith\OneDrive\Documentos\Universidad\Máster IA\T1-herramientas_programación\notebooks\data\2014-baby-names-raw.csv", r"C:\Users\edith\OneDrive\Documentos\Universidad\Máster IA\T1-herramientas_programación\notebooks\data\2015-baby-names-raw.csv"]
read_baby_names(paths)
```
